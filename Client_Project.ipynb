```python
query = """
SELECT 
    c.FullName, 
    c.PhoneNumber, 
    c.Email, 
    o.TotalCost
FROM CustomersDetails c
JOIN Bookings b ON c.CustomerID = b.`Customers Detail_CustomerID`
JOIN Orders o ON b.BookingID = o.Bookings_BookingID
WHERE o.TotalCost > 60;
"""

cursor.execute(query)

# Fetch results
results = cursor.fetchall()
```

```python
print("Customers who spent more than $60:\n")
for row in results:
    print(f"Name: {row[0]}, Phone: {row[1]}, Email: {row[2]}, Bill: ${row[3]}")
```
